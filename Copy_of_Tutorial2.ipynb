# 1. Create a numpy array of size 2 x 2

#numpy is a python library that allows you to work with numbers and large datasets etc

```python
import numpy as np

array_2x2 = np.array ([
    [1, 2],
    [3, 4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

# 2. Create a random numpy array of size 3 x 4

```python
random3x4 = np.random.rand(3, 4)

print("Random 3x4 array")
print(random3x4)
```

# 3. Create a numpy array of ones and a numpy array of zeroes

```python
ones_array = np.ones((3, 3))
zeros_array = np.zeros((3, 3))

print("Array of Ones:")
print(ones_array)

print("Array of zeros:")
print(zeros_array)
```

# 4. Create a random numpy array of size 6x3 and transpose it

```python
random6x3 = np.random.rand(6, 3)
transposed_array = random6x3.T

print("Random 6x3 array:")
print(random6x3)

print("Transposed array (3x6):")
print(transposed_array)
```

# 5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n, p, m = 2, 3, 4

array_np = np.random.rand(n, p)
array_pm = np.random.rand(p, m)
dot_product = np.dot(array_np, array_pm)

print("First array (n x p):")
print(array_np)

print("Second array (p x m):")
print(array_pm)

print("Dot product (n x m):")
print(dot_product)
```
